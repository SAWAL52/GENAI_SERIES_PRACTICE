### Prompt Templates- Dynamic prompts

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOllama(
    model="llama3.2:3b",
    base_url="http://127.0.0.1:11434"
)
```

```python
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator from English to {language}"),
    ("user", "Translate: {query}")
])
```

```python
final_prompts= prompts.format_messages(language="Chinese",query="China is second largest economy of the world.")
final_prompts
```

```python
response=llm.invoke(final_prompts)
print(response.content)
```

### Chains- Runnable

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(
    model="llama3.2:3b",
    base_url="http://127.0.0.1:11434"
)
out = StrOutputParser()

prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator from English to {language}"),
    ("user", "Translate: {query}")
])
```

```python
#response=llm.invoke(prompts.invoke({"language":"Hindi","query":"I love you."})
#print(response.content)
```

```python
def transform_case(result:str):
    return result.upper()
```

```python
chains= prompts | llm | out | transform_case
chains
```

```python
response=chains.invoke({"language":"Hindi","query":"I love LangChain."})
print(response)
```
