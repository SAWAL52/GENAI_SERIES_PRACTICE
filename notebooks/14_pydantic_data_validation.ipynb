```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name: str 
    email: str 
    age: int 
```

```python
user= User(name="Sawal",email="sharmasawal3@gmail.com",age=23)
user
```

### Field Validation

```python
class Product(BaseModel):
    title:str=Field(max_length=25)
    description:str=Field(min_length=15)
    price:int=Field(gt=0)
    
    
```

```python
product= Product(title="Iphone 17 Pro",description="Latest Apple Phone released in 2025",price=221300)
product
```

```python
class Order(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    product:Product
```

```python
order=Order(id="et3294",product=product)
order.product.title, order.product.price
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items: List[Product]
    comment: str= Field(default="No Comment")
```

```python
cart=Carts(id="nt29394",items=[product])
cart
```

```python
for product in cart.items:
    print(product.title)
```

#### Annotated

```python
commonValidator=Annotated[List[Product],Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items: List[Product]
    comment: str= Field(default="No Comment")
```

```python
class Order(BaseModel):
    products:commonValidator
```

```python
order=Order()
order
```
