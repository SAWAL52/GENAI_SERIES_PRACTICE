```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver


```

```python
model = ChatGroq(model="openai/gpt-oss-20b",streaming="True")
search = GoogleSerperAPIWrapper()
```

```python
search.run("Address of St. Xavier's College in Nepal?") #Example
```

```python
agent=create_agent(
    model = model,
    tools=[search.run],
    system_prompt="You are a agent and can search any question on google.",
    checkpointer=MemorySaver()
)
```

```python
question="Who won US presidential elections in 2024?"
```

```python
response=agent.invoke({"messages":[{"role":"user", "content":question}]},
                      {"configurable":{"thread_id":"1"}})

print(response["messages"][-1].content)
```
