```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int):
    """
        it will return the sum of two numbers

        Args:
            a: number one
            b: number two
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int):
    """
        it will return the product of two numbers

        Args:
            a: number one
            b: number two
    """
    return a * b
```

```python
add_numbers.invoke({"a":10, "b":15})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent

llm = ChatGroq(model="openai/gpt-oss-20b",streaming="True")

```

```python
agent= create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a math teacher, and always use tools for calculation."
)
```

```python
response=agent.invoke({"messages":[{"role":"user", "content":"What is 5+8? "}]})

print(response["messages"][-1].content)
```

```python
response=agent.invoke({"messages":[{"role":"user", "content":"What is 5*8? "}]})

print(response["messages"][-1].content)
```

```python
response=agent.invoke({"messages":[{"role":"user", "content":"What is product of 2 and 3?"}]})

print(response["messages"][-1].content)
```
