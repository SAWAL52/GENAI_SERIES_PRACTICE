### Load Env Variables


```python
from dotenv import load_dotenv

load_dotenv()
```

### Connect with ChatOllama model

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="llama3.2:3b",
    base_url="http://127.0.0.1:11434"
)
```

### Questions and Prompts

```python

response = llm.invoke("What is the height of Mt. Everest?")
print(response.content)
```

```python

prompts = [
    ("system", "You are a Python developer."),
    ("user", "Main logic of Dijkstra's algorithm")
]

response = llm.invoke(prompts)

print(response.content)
```
